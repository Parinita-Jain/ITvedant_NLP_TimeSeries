# Auto regression example

```python
# Auto Regression example - suitable for data without trend and seasonal component
from statsmodels.tsa.ar_model import AutoReg

# create a linear data
data = [10,12,14,16,18,20,20,22] #linear data
```

```python
import matplotlib.pyplot as plt
plt.plot(data)
```

```python
ARmodel = AutoReg(data, lags=1) #calling Auto regression model
ARmodel_fit = ARmodel.fit() #Fitting model
```

```python
# making predictions
ypredicted = ARmodel_fit.predict(8,10)
print(ypredicted)
```

```python
pred_list = ypredicted.tolist()
combinedlist = data+pred_list
```

```python
plt.plot(combinedlist)
```

# Moving average example

```python
from statsmodels.tsa.arima.model import ARIMA #suitable for data without trend and seasonality
```

```python
import matplotlib.pyplot as plt
plt.plot(data)
```

```python
# fit model
MAmodel = ARIMA(data, order=(0,0, 1)) #model with AR=0 and MA=1
MAmodel_fit = MAmodel.fit()
```

```python
# make prediction
ypredicted = MAmodel_fit.predict(8,10)
print(ypredicted)
```

```python
pred_list = ypredicted.tolist()
combinedlist = data+pred_list
```

```python
plt.plot(combinedlist)
```

# Autoregressive Moving Average (ARMA)

```python
# Import Data -  Australian Drug Sales
import pandas as pd
DrugSalesData = pd.read_csv('/content/AusAntidiabeticDrug.csv', parse_dates=['Date'], index_col='Date')
```

```python
import matplotlib.pyplot as plt
plt.rcParams.update({'figure.figsize': (10,6)})
plt.plot(DrugSalesData['Value'])
```

```python
# fit model
from statsmodels.tsa.arima.model import ARIMA
ARMAmodel = ARIMA(DrugSalesData['Value'], order=(1, 0, 1))
ARmodel_fit = ARMAmodel.fit()
```

```python
# make prediction
ypredicted = ARmodel_fit.predict(len(DrugSalesData), len(DrugSalesData)+2)
print(ypredicted)
```

# Autoregressive Integrated Moving Average (ARIMA)

```python
from statsmodels.tsa.arima_model import ARIMA
```

```python
# fit model
from statsmodels.tsa.arima.model import ARIMA
ARIMAmodel = ARIMA(DrugSalesData['Value'], order=(1, 1, 1)) #notice p,d and q value here
ARIMA_model_fit = ARIMAmodel.fit()

# make prediction
ypredicted = ARIMA_model_fit.predict(len(DrugSalesData), len(DrugSalesData)+2)
print(ypredicted)
```
